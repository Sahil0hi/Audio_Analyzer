```python
import numpy as np
import sys
import math
import pandas as pd


Rvals = [10000,12000,15000,18000,22000,27000,33000,39000,47000,56000,68000,82000,100000,
         120000,150000,180000,220000,270000,330000,390000,470000,560000,680000,820000,1000000]
CVals = [
          10*10**-12,      #10pF
          47*10**-12,      #47pF
          0.1*10**-9,      #0.1nF
          0.22*10**-9,     #0.22nF
          0.33*10**-9,     #0.33nF
          0.47*10**-9,     #0.47nF
          0.68*10**-9,     #0.68nF
          1*10**-9,        #1nF
          1.5*10**-9,      #1.5nF
          2.2*10**-9,      #2.2nF
          3.3*10**-9,      #3.3nF
          4.7*10**-9,      #4.7nF
          6.8*10**-9,      #6.8nF
          10*10**-9,       #10nF
          15*10**-9,       #15nF
          22*10**-9,       #22nF
          33*10**-9,       #33nF
          47*10**-9,       #47nF
          68*10**-9,       #68nF
          0.1*10**-6,      #0.1uF
          0.15*10**-6,     #0.15uF
          0.47*10**-6,     #0.47uF
          1*10**-6,        #1uF
          4.7*10**-6       #4.7uF
         ]
```

```python
import math

#This code selects values for high pass filter
#These are the functions for calculating your high pass cutoff frequency and Quality Factor
#Use the functions in the Sallen-Kay Architecture pdf, and finish these functions to calculate the cutoff frequncy and quality factor given a set of component values
def CutoffFrequency(C1, C2, R1, R2):
  #TODO finish this function
  fc = 1/(2*math.pi * math.sqrt(R1*R2*C1*C2))

  return(fc)

def Quality_Factor(C1, C2, R1, R2):
  #TODO finish this function
  Q = math.sqrt(R1 * R2 * C1 * C2) / (R1 * (C1 + C2))
  return Q


#To determine if a set of components creates an acceptable circuit, we will compare the frequency and quality we calculate with the values we want
#TO do this, we'll set an upper and lower bound that values we deem 'acceptable' fall between

freqdes = 700 #TODO make this your desired high pass cutoff frequency
#This tolerance is used to set the high and low bound. For example, 0.1*freqdes will save values within 10% of your desired frequency
ftolerance = 0.10 * freqdes #TODO adjust the tolerance to narrow down your solutions
fmax = freqdes + ftolerance
fmin = freqdes - ftolerance

Qdes = 0.707 #TODO make this your desired quality factor
#This tolerance is used to set the high and low bound. For example, 0.1*Qdes will save values within 10% of your desired cutoff
Qtolerance = 1 * Qdes #TODO adjust the tolerance to narrow down your solutions
Qmax = Qdes + Qtolerance
Qmin = Qdes - Qtolerance


#Creates and array that all the valid solutions will be saved to and printed
solutions = []

#Defines each component to be an array of every possible value
R1 = Rvals
R2 = Rvals
C1 = CVals
C2 = CVals

#This loops through an array for every value, and checks the corner frequency and quality factor for every possible combination of component values
#If a set of values produces a corner frequency and quality within your chosen tolerance, it will print out that set of values
for r in range(len(R1)):
  for R in range(len(R2)):
    for c in range(len(C1)):
      for C in range(len(C2)):
        #Calculates the corner and quality for the current set of values
        corner = CutoffFrequency(C1[c], C2[C], R1[r], R2[R])
        quality = Quality_Factor(C1[c], C2[C], R1[r], R2[R])

        #If the corner and cutoff that were calculated are within your bounds, they're appended to the solution array
        if corner <= fmax and corner >= fmin and quality <= Qmax and quality >= Qmin:
          solutions.append(np.array([C1[c], C2[C], R1[r], R2[R], corner, quality]))


#Puts the soluions into a dataframe to read
df = pd.DataFrame(solutions, columns=['C1', 'C2', 'R1', 'R2', 'Corner f', 'Quality Factor'])
print(df)
```
